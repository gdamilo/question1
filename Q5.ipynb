```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Manually inserted PCA-transformed training data (example values for 3 principal components)
X_train_pca = [
    [2.5, 0.7, 0.1],  # First training sample PCA values
    [1.9, 0.5, 0.2],  # Second training sample PCA values
    [3.1, 0.6, 0.3],  # Third training sample PCA values
    [2.0, 0.4, 0.4],  # Fourth training sample PCA values
    [2.2, 0.8, 0.5]   # Fifth training sample PCA values
]

# Manually inserted PCA-transformed testing data (example values for 3 principal components)
X_test_pca = [
    [2.4, 0.6, 0.2],  # First test sample PCA values
    [2.3, 0.7, 0.3]   # Second test sample PCA values
]

# Manually inserted target labels for training and testing
y_train = [1, 0, 1, 0, 1]  # Example training labels
y_test = [0, 1]            # Example testing labels

# Initialize Logistic Regression model
log_reg_pca = LogisticRegression(max_iter=1000)

# Train Logistic Regression model on the manually inserted PCA-transformed data
log_reg_pca.fit(X_train_pca, y_train)

# Predict on the test set using the PCA-transformed test data
y_pred_pca = log_reg_pca.predict(X_test_pca)

# Calculate performance metrics
accuracy_pca = accuracy_score(y_test, y_pred_pca)
precision_pca = precision_score(y_test, y_pred_pca)
recall_pca = recall_score(y_test, y_pred_pca)
f1_pca = f1_score(y_test, y_pred_pca)

# Print the metrics
print(f"Logistic Regression with PCA - Accuracy: {accuracy_pca:.4f}")
print(f"Precision: {precision_pca:.4f}")
print(f"Recall: {recall_pca:.4f}")
print(f"F1 Score: {f1_pca:.4f}")

# Confusion matrix
cm_pca = confusion_matrix(y_test, y_pred_pca)
ConfusionMatrixDisplay(cm_pca).plot()
plt.title("Confusion Matrix - Logistic Regression with PCA")
plt.show()
```
